```python
import numpy as np
from scipy.misc import derivative

def empirical_mgf(t, data):
    """ Compute the empirical MGF at a specific t using data. """
    return np.mean(np.exp(t * data))

def estimate_moments(data):
    """ Estimate the first and second moments using numerical differentiation of the MGF. """
    # Point at which to estimate the derivatives
    t0 = 0  # We differentiate at t=0
    
    # First derivative to get the first moment (mean)
    first_moment = derivative(lambda t: empirical_mgf(t, data), t0, dx=1e-6, n=1)
    
    # Second derivative to get the second moment
    second_moment = derivative(lambda t: empirical_mgf(t, data), t0, dx=1e-6, n=2)
    
    return first_moment, second_moment

# Example data
data = np.random.normal(loc=0, scale=1, size=30000)  # Simulate some normal data

# Estimate moments
first_moment, second_moment = estimate_moments(data)
print("Estimated First Moment (Mean):", first_moment)
print("Estimated Second Moment:", second_moment)
```
