```python
import numpy as np
import matplotlib.pyplot as plt

# Define the number of terms and initial conditions
num_terms = 50
a = np.zeros(num_terms + 1)  # Array to hold coefficients from a_0 to a_100
a[0] = 1  # Initial condition a_0
a[1] = 0  # Initial condition a_1

# Compute coefficients using the recurrence relation
for n in range(num_terms - 1):  # n goes from 0 to 98
    a[n+2] = (3 * (n + 1) * a[n+1] - 2 * a[n]) / ((n + 2) * (n + 1))

# Generate a range of x values
x_values = np.linspace(0, 2*np.pi, 400)  # x values from -1 to 1
# Compute the polynomial value at each x
polynomial_values = np.polyval(a[::-1], x_values)  # np.polyval expects coefficients from highest to lowest degree

# Plot the polynomial
plt.figure(figsize=(10, 6))
plt.plot(x_values, polynomial_values, label='Polynomial $p(x)$')
plt.plot(x_values,-1*np.exp(2*x_values)+2*np.exp(x_values),'r--',label='analytical')
plt.title('Plot of the Polynomial $p(x) = \sum_{n=0}^{100} a_n x^n$')
plt.xlabel('x')
plt.ylabel('$p(x)$')
plt.grid(True)
plt.legend()
plt.show()
```
