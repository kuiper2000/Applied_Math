```python
import numpy as np
import time

# Generate two random time series with a dimension of 100000
np.random.seed(42)  # For reproducibility
time_series_1 = np.random.randn(50000)
time_series_2 = np.random.randn(50000)

# Measure time correctly
start_time = time.time()
convolution_result = np.convolve(time_series_1, time_series_2, mode='full')
end_time = time.time()

# Print results
print("Size of convolution result:", convolution_result.shape)
print(f"Time taken: {end_time - start_time:.6f} seconds")
```

```python
import numpy as np
from scipy.signal import fftconvolve

# Generate two random time series with a dimension of 100000
np.random.seed(42)  # For reproducibility

time_series_1 = np.random.randn(50000)
time_series_2 = np.random.randn(50000)
# Measure time correctly
start_time = time.time()
# Compute the convolution using scipy.signal.fftconvolve
convolution_result = fftconvolve(time_series_1, time_series_2, mode='full')
end_time = time.time()
# Print the size of the result
print("Size of convolution result:", convolution_result.shape)
print(f"Time taken: {end_time - start_time:.6f} seconds")
```
